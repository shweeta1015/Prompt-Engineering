```python
import os
from google import genai
from google.genai import types

# Ensure your API key is set before running this cell
if 'GEMINI_API_KEY' not in os.environ:
    # You can set your API key directly here if it's not in your environment variables:
    # os.environ['GEMINI_API_KEY'] = 'YOUR_API_KEY_HERE'
    print('Please ensure GEMINI_API_KEY is set in your environment.')

client = genai.Client()
```

## Standard Generation with Config
Demonstrating `temperature`, `max_output_tokens`, `stop_sequences`.

```python
response = client.models.generate_content(
    model="gemini-3.5-flash",
    contents="Explain how photosynthesis works in 3 short sentences.",
    config=types.GenerateContentConfig(
        temperature=0.7,
        max_output_tokens=200,
        stop_sequences=["END"],
        # response_logprobs and logprobs are commented out because they are
        # not supported by gemini-3.5-flash (and gemini-2.5-flash is deprecated).
        # response_logprobs=True,
        # logprobs=5,
    ),
)
print(response.text)
```

## Streaming Generation
Using `generate_content_stream` to stream the response.

```python
for chunk in client.models.generate_content_stream(
    model="gemini-3.5-flash",
    contents="Write a short poem about the ocean",
):
    print(chunk.text, end="")
print()
```
